Data Preparation and Pre-processing

```python
import pandas as pd
import numpy as np
import re
import string
import nltk
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split

# Download NLTK stop-words
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

print("NLTK stopwords loaded. Total English stop-words:", len(stop_words))
```

```python
# 1. Load dataset
df = pd.read_csv("US-Economic-News.csv", encoding="latin-1")

print(f"Raw dataset shape: {df.shape}")
print("Columns present:", list(df.columns))

# Identify text and target columns
# We combine 'headline' and 'text' to form the full textual content
df_clean = df.drop(columns=df.columns.difference(['relevance', 'headline', 'text'])).copy()
df_clean['combined_text'] = df_clean['headline'] + ' ' + df_clean['text']

# Filter out ambiguous annotations ('not sure')
df_clean = df_clean[df_clean['relevance'] != 'not sure'].copy()

print(f"\nDataset shape after filtering 'not sure': {df_clean.shape}")
print("\nRelevance class distribution (overall):")
print(df_clean['relevance'].value_counts())

# 2. Stratified train-test split (80% train, 20% test)
X = df_clean['combined_text']
y = df_clean['relevance']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(f"\nInitial Split -> Training: {len(X_train)}, Testing: {len(X_test)}")
```

```python
# 3. Text Pre-processing functions
def remove_unnecessary_characters(text):
    """Removes HTML tags, HTML entities, and non-ASCII characters/artifacts."""
    text = re.sub(r'<.*?>', ' ', text)          # Strip HTML tags like </br>
    text = re.sub(r'&[a-z]+;', ' ', text)        # Strip HTML entities like &amp;, &nbsp;
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)   # Strip non-ASCII symbols and encoding artifacts
    return text

def remove_digits(text):
    """Removes all numeric digits."""
    return re.sub(r'\d+', ' ', text)

def remove_punctuation(text):
    """Removes punctuation characters."""
    return text.translate(str.maketrans('', '', string.punctuation))

def remove_whitespace(text):
    """Collapses multiple whitespace characters and strips leading/trailing spaces."""
    return re.sub(r'\s+', ' ', text).strip()

def clean_pipeline(text):
    """Complete modular text pre-processing pipeline."""
    text = text.lower()                                                # 1. Lowercase
    text = remove_unnecessary_characters(text)                         # 2. HTML / entities / non-ASCII
    text = remove_digits(text)                                         # 3. Digits
    text = remove_punctuation(text)                                     # 4. Punctuation
    tokens = [word for word in text.split() if word not in stop_words and len(word) > 1] # 5. Stop-words & single chars
    text = ' '.join(tokens)
    text = remove_whitespace(text)                                      # 6. Whitespace normalization
    return text

# Apply pre-processing to train and test sets
X_train_clean = X_train.apply(clean_pipeline)
X_test_clean = X_test.apply(clean_pipeline)

print("Pre-processing pipeline applied to training and testing sets successfully.")
```

```python
# 4. Report Training & Testing instances and distributions of Relevant / Not Relevant
print("=" * 65)
print("DATA PREPARATION & PRE-PROCESSING REPORT")
print("=" * 65)
print(f"Total Cleaned Dataset Size: {len(df_clean)} articles")
print(f"Training instances:         {len(X_train_clean)} ({len(X_train_clean)/len(df_clean)*100:.2f}%)")
print(f"Testing instances:          {len(X_test_clean)} ({len(X_test_clean)/len(df_clean)*100:.2f}%)")

train_dist = pd.DataFrame({
    'Count': y_train.value_counts(),
    'Percentage': (y_train.value_counts(normalize=True) * 100).round(2).astype(str) + '%'
})
print("\n--- Training Set Class Distribution ---")
display(train_dist)

test_dist = pd.DataFrame({
    'Count': y_test.value_counts(),
    'Percentage': (y_test.value_counts(normalize=True) * 100).round(2).astype(str) + '%'
})
print("\n--- Testing Set Class Distribution ---")
display(test_dist)
```

```python
# 5. Preprocessing Verification: Sample Comparisons and Step-by-Step Breakdown
sample_idx = 0
sample_raw = X_train.iloc[sample_idx]
sample_clean = X_train_clean.iloc[sample_idx]

print("=" * 75)
print("SAMPLE COMPARISON (BEFORE VS. AFTER PRE-PROCESSING)")
print("=" * 75)
print("[BEFORE PRE-PROCESSING] (First 350 characters):")
print(sample_raw[:350] + "...")
print("\n[AFTER PRE-PROCESSING] (First 350 characters):")
print(sample_clean[:350] + "...")

# Step-by-step breakdown demonstration
s0 = sample_raw
s1 = s0.lower()
s2 = remove_unnecessary_characters(s1)
s3 = remove_digits(s2)
s4 = remove_punctuation(s3)
s5 = ' '.join([w for w in s4.split() if w not in stop_words and len(w) > 1])
s6 = remove_whitespace(s5)

print("\n" + "=" * 75)
print("STEP-BY-STEP TRANSFORMATION BREAKDOWN")
print("=" * 75)
print("0. Raw Text:               ", s0[:110], "...")
print("1. Lowercased:             ", s1[:110], "...")
print("2. HTML/Entities Removed:  ", s2[:110], "...")
print("3. Digits Removed:         ", s3[:110], "...")
print("4. Punctuation Removed:    ", s4[:110], "...")
print("5. Stop-Words Removed:     ", s5[:110], "...")
print("6. Whitespace Normalized:  ", s6[:110], "...")
```

Feature Extraction

```python
from sklearn.feature_extraction.text import CountVectorizer

feature_settings = [40000, 5000, 1000]
vectorizers = {}
X_train_bow = {}
X_test_bow = {}
rep_summary = []

for mf in feature_settings:
    # Initialize CountVectorizer with specified max_features
    vec = CountVectorizer(max_features=mf)
    
    # Fit strictly on training set, transform on train and test
    X_tr = vec.fit_transform(X_train_clean)
    X_te = vec.transform(X_test_clean)
    
    vectorizers[mf] = vec
    X_train_bow[mf] = X_tr
    X_test_bow[mf] = X_te
    
    vocab_size = len(vec.vocabulary_)
    n_rows, n_cols = X_tr.shape
    non_zeros = X_tr.nnz
    total_cells = n_rows * n_cols
    sparsity_pct = (1.0 - (non_zeros / total_cells)) * 100.0
    
    # Memory consumption
    sparse_mem_mb = (X_tr.data.nbytes + X_tr.indices.nbytes + X_tr.indptr.nbytes) / (1024 * 1024)
    dense_mem_mb = (n_rows * n_cols * 8) / (1024 * 1024)  # 64-bit integer representation
    savings_pct = (1.0 - (sparse_mem_mb / dense_mem_mb)) * 100.0
    
    rep_summary.append({
        'max_features Setting': mf,
        'Actual Vocab Size': vocab_size,
        'Train Matrix Shape': f"{n_rows} x {n_cols}",
        'Non-Zero Elements': non_zeros,
        'Sparsity (%)': round(sparsity_pct, 3),
        'Sparse Memory (MB)': round(sparse_mem_mb, 2),
        'Dense Memory (MB)': round(dense_mem_mb, 2),
        'Memory Savings (%)': round(savings_pct, 2)
    })

df_rep_summary = pd.DataFrame(rep_summary)
print("=" * 75)
print("FEATURE EXTRACTION & REPRESENTATION SIZE COMPARISON")
print("=" * 75)
display(df_rep_summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Actual Vocabulary Size
sns.barplot(data=df_rep_summary, x='max_features Setting', y='Actual Vocab Size', ax=axes[0], palette='Blues_r')
axes[0].set_title('Vocabulary Size (|V|)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('max_features Setting')
axes[0].set_ylabel('Vocabulary Count')
for p in axes[0].patches:
    axes[0].annotate(f"{int(p.get_height()):,}", (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                     ha='center', va='center', color='white', fontweight='bold')

# 2. Non-Zero Entries
sns.barplot(data=df_rep_summary, x='max_features Setting', y='Non-Zero Elements', ax=axes[1], palette='Greens_r')
axes[1].set_title('Non-Zero Elements in Training Matrix', fontsize=13, fontweight='bold')
axes[1].set_xlabel('max_features Setting')
axes[1].set_ylabel('Non-Zero Count')
for p in axes[1].patches:
    axes[1].annotate(f"{int(p.get_height()):,}", (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                     ha='center', va='center', color='white', fontweight='bold')

# 3. Sparse Memory Footprint (MB)
sns.barplot(data=df_rep_summary, x='max_features Setting', y='Sparse Memory (MB)', ax=axes[2], palette='Oranges_r')
axes[2].set_title('Sparse Matrix Memory Footprint (MB)', fontsize=13, fontweight='bold')
axes[2].set_xlabel('max_features Setting')
axes[2].set_ylabel('RAM (MB)')
for p in axes[2].patches:
    axes[2].annotate(f"{p.get_height():.2f} MB", (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                     ha='center', va='center', color='white', fontweight='bold')

plt.suptitle('Comparison of Feature Representation Size across max_features Settings', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

Classification and Model Comparison

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

# Factory for clean classifier initialization
classifier_factories = {
    "Multinomial Naive Bayes": lambda: MultinomialNB(),
    "Logistic Regression": lambda: LogisticRegression(max_iter=1000, random_state=42),
    "LinearSVC (balanced)": lambda: LinearSVC(class_weight="balanced", random_state=42, max_iter=2000)
}

results_data = []
confusion_matrices = {}
classification_reports = {}

for clf_name, clf_init in classifier_factories.items():
    for mf in feature_settings:
        # Initialize new classifier
        clf = clf_init()
        
        # Retrieve corresponding BoW features
        X_tr = X_train_bow[mf]
        X_te = X_test_bow[mf]
        
        # Train model
        clf.fit(X_tr, y_train)
        
        # Predict on unseen test set
        y_pred = clf.predict(X_te)
        
        # Metrics for the positive class ('yes' = Relevant)
        acc = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, pos_label='yes', zero_division=0)
        rec = recall_score(y_test, y_pred, pos_label='yes', zero_division=0)
        f1 = f1_score(y_test, y_pred, pos_label='yes', zero_division=0)
        macro_f1 = f1_score(y_test, y_pred, average='macro')
        
        # Confusion Matrix: rows = actual, columns = predicted (labels: ['no', 'yes'])
        cm = confusion_matrix(y_test, y_pred, labels=['no', 'yes'])
        
        config_key = f"{clf_name} | max_features={mf}"
        confusion_matrices[config_key] = cm
        classification_reports[config_key] = classification_report(y_test, y_pred, digits=4)
        
        results_data.append({
            'Classifier': clf_name,
            'max_features': mf,
            'Accuracy': round(acc, 4),
            'Precision (yes)': round(prec, 4),
            'Recall (yes)': round(rec, 4),
            'F1-Score (yes)': round(f1, 4),
            'Macro F1-Score': round(macro_f1, 4),
            'True Neg (no)': cm[0, 0],
            'False Pos (no->yes)': cm[0, 1],
            'False Neg (yes->no)': cm[1, 0],
            'True Pos (yes)': cm[1, 1]
        })

df_results = pd.DataFrame(results_data)
print("All 9 configurations trained and evaluated successfully!")
```

```python
# Display comprehensive metrics table
print("=" * 85)
print("STAGE 3: COMPREHENSIVE CLASSIFIER EVALUATION TABLE")
print("=" * 85)

# Display sorted by Classifier and max_features
display(df_results.sort_values(by=['Classifier', 'max_features'], ascending=[True, False]))

print("\n--- Top Models Ranked by F1-Score (Relevant class) ---")
display(df_results.sort_values(by='F1-Score (yes)', ascending=False).head(5))
```

```python
# 3x3 Confusion Matrix Heatmap Grid
fig, axes = plt.subplots(3, 3, figsize=(18, 16))
clf_names = list(classifier_factories.keys())

for row_idx, clf_name in enumerate(clf_names):
    for col_idx, mf in enumerate(feature_settings):
        ax = axes[row_idx, col_idx]
        config_key = f"{clf_name} | max_features={mf}"
        cm = confusion_matrices[config_key]
        
        # Calculate percentages per actual class (row normalization)
        cm_pct = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] * 100.0
        annot_labels = np.asarray([
            f"{count}\n({pct:.1f}%)"
            for count, pct in zip(cm.flatten(), cm_pct.flatten())
        ]).reshape(2, 2)
        
        sns.heatmap(
            cm, annot=annot_labels, fmt="", cmap="Blues", cbar=False, ax=ax,
            xticklabels=["Pred: No", "Pred: Yes"],
            yticklabels=["Actual: No", "Actual: Yes"],
            annot_kws={"fontsize": 11, "fontweight": "bold"}
        )
        ax.set_title(f"{clf_name}\nmax_features={mf:,}", fontsize=12, fontweight='bold')
        ax.set_xlabel("Predicted Label")
        ax.set_ylabel("True Label")

plt.suptitle("3x3 Confusion Matrix Grid: All 9 Classifier & Feature Combinations", fontsize=16, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```
