```python
import pandas as pd
import re
import string
import nltk
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split

#Download stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

#Load dataset
df = pd.read_csv("US-Economic-News.csv", encoding="latin-1")


df_clean = df.drop(columns=df.columns.difference(['relevance', 'headline', 'text'])).copy()
df_clean['combined_text'] = df_clean['headline'] + ' ' + df_clean['text']
df_clean = df_clean[df_clean['relevance'] != 'not sure'].copy()


X = df_clean['combined_text']
y = df_clean['relevance']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)


def remove_unnecessary_characters(text):
    text = re.sub(r'<.*?>', ' ', text)          # Strip HTML tags
    text = re.sub(r'&[a-z]+;', ' ', text)        # Strip HTML entities
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)   # Strip non-ASCII symbols
    return text

def remove_digits(text):
    return re.sub(r'\d+', ' ', text)

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

def remove_whitespace(text):
    return re.sub(r'\s+', ' ', text).strip()

def clean_pipeline(text):
    text = text.lower()                                                # Lowercase
    text = remove_unnecessary_characters(text)                         # HTML / special chars
    text = remove_digits(text)                                         # Digits
    text = remove_punctuation(text)                                     # Punctuation
    tokens = [word for word in text.split() if word not in stop_words] # Stopwords
    text = ' '.join(tokens)
    text = remove_whitespace(text)                                      # Whitespace
    return text


X_train_clean = X_train.apply(clean_pipeline)
X_test_clean = X_test.apply(clean_pipeline)


print("Pre-processing Complete!")
print("-" * 40)
print(f"Training instances: {len(X_train_clean)}")
print(f"Testing instances:  {len(X_test_clean)}")
print("\nClass distribution in Training set:")
print(y_train.value_counts())
print("\nClass distribution in Testing set:")
print(y_test.value_counts())
```

```python
sample_raw = X_train.iloc[0]

step_lower = sample_raw.lower()
step_no_html = remove_unnecessary_characters(step_lower)
step_no_digits = remove_digits(step_no_html)
step_no_punct = remove_punctuation(step_no_digits)
step_no_stopwords = ' '.join([w for w in step_no_punct.split() if w not in stop_words and len(w) > 1])
step_final = remove_whitespace(step_no_stopwords)

print("STEP-BY-STEP PREPROCESSING BREAKDOWN:")
print("0. Raw Text:                  ", sample_raw[:120], "...")
print("1. Lowercasing:               ", step_lower[:120], "...")
print("2. HTML/Artifact Removal:     ", step_no_html[:120], "...")
print("3. Digits Removed:            ", step_no_digits[:120], "...")
print("4. Punctuation Removed:       ", step_no_punct[:120], "...")
print("5. Stop-Words Removed:        ", step_no_stopwords[:120], "...")
print("6. Whitespace Normalized:     ", step_final[:120], "...")
```
